```python
MODELS = {
    # LoRA models
    "BART_LoRA": {"name": "facebook/bart-large-cnn", "lora": True},
    "T5_LoRA": {"name": "t5-base", "lora": True},
    "FLAN_T5_LoRA": {"name": "google/flan-t5-base", "lora": True},
    "MBART_LoRA": {"name": "facebook/mbart-large-50", "lora": True},

    #  Normal models
    "PEGASUS": {"name": "google/pegasus-cnn_dailymail", "lora": False},
    "DistilBART": {"name": "sshleifer/distilbart-cnn-12-6", "lora": False},
    "BART_BASE": {"name": "facebook/bart-base", "lora": False},
    "T5_SMALL": {"name": "t5-small", "lora": False}
}
```

```python
def apply_lora(model, model_name):

    if "bart" in model_name:
        target_modules = ["q_proj", "v_proj"]

    elif "t5" in model_name or "flan" in model_name:
        target_modules = ["q", "v"]

    elif "mbart" in model_name:
        target_modules = ["q_proj", "v_proj"]

    else:
        return model

    config = LoraConfig(
        r=8,
        lora_alpha=16,
        target_modules=target_modules,
        lora_dropout=0.1,
        bias="none"
    )

    return get_peft_model(model, config)
```

```python
model_cache = {}

def load_model(key, cfg):

    if key in model_cache:
        return model_cache[key]

    tokenizer = AutoTokenizer.from_pretrained(cfg["name"])
    model = AutoModelForSeq2SeqLM.from_pretrained(cfg["name"]).to(device)

    if cfg["lora"]:
        model = apply_lora(model, cfg["name"])

    model_cache[key] = (tokenizer, model)

    return tokenizer, model
```

```python
def generate_summary(model, tokenizer, article, model_name):

    if "t5" in model_name:
        article = "summarize: " + article

    inputs = tokenizer(
        article,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    ).to(device)

    output = model.generate(
        **inputs,
        max_length=128,
        min_length=50,
        num_beams=5,
        length_penalty=2.0,
        early_stopping=True
    )

    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
def run_all_models(article, reference):

    for key, cfg in MODELS.items():

        print("\n" + "="*70)
        print(f"MODEL: {key}")

        tokenizer, model = load_model(key, cfg)

        summary = generate_summary(model, tokenizer, article, cfg["name"])

        print("\n Summary:")
        print(summary)

        # BLEU
        bleu_score = bleu.compute(
            predictions=[summary],
            references=[[reference]]
        )["bleu"]

        # ROUGE
        scores = rouge.compute(
            predictions=[summary],
            references=[reference]
        )

        print("\nMetrics:")
        print(f"BLEU: {bleu_score:.4f}")
        print(f"ROUGE-1: {scores['rouge1']:.4f}")
        print(f"ROUGE-2: {scores['rouge2']:.4f}")
        print(f"ROUGE-L: {scores['rougeL']:.4f}")
```

```python
import ipywidgets as widgets
from IPython.display import display
from google.colab import output

output.enable_custom_widget_manager()
```

```python
article_box = widgets.Textarea(
    placeholder="Enter article...",
    layout=widgets.Layout(width='100%', height='150px')
)

reference_box = widgets.Textarea(
    placeholder="Enter reference summary...",
    layout=widgets.Layout(width='100%', height='100px')
)

button = widgets.Button(description="Run Models")

output = widgets.Output()

def on_click(b):
    with output:
        output.clear_output()

        article = article_box.value
        reference = reference_box.value if reference_box.value else article[:150]

        run_all_models(article, reference)

button.on_click(on_click)

display(article_box, reference_box, button, output)
```

```python
def compare_models(article, reference):

    results = []

    for key, cfg in MODELS.items():

        print(f" Running {key}")

        tokenizer, model = load_model(key, cfg)

        summary = generate_summary(model, tokenizer, article, cfg["name"])

        # BLEU
        bleu_score = bleu.compute(
            predictions=[summary],
            references=[[reference]]
        )["bleu"]

        # ROUGE
        scores = rouge.compute(
            predictions=[summary],
            references=[reference]
        )

        results.append({
            "Model": key,
            "BLEU": round(bleu_score, 4),
            "ROUGE-1": round(scores["rouge1"], 4),
            "ROUGE-2": round(scores["rouge2"], 4),
            "ROUGE-L": round(scores["rougeL"], 4)
        })

    df = pd.DataFrame(results)

    # Sort by ROUGE-L (best metric)
    df = df.sort_values(by="ROUGE-L", ascending=False)

    return df
```

```python
article = """India’s economy is showing signs of recovery with increased manufacturing output and consumer demand."""

reference = """India’s economy is recovering with improved manufacturing and demand."""

df_results = compare_models(article, reference)

df_results
```
